```python
import requests
from bs4 import BeautifulSoup
```

```python
# Function to fetch hyperlinks from a given URL
def fetch_hyperlinks(url):
    try:
        # Send HTTP request to the URL
        response = requests.get(url)
        response.raise_for_status()  # Raise an error for bad status codes

        # Parse the HTML content
        soup = BeautifulSoup(response.text, 'html.parser')

        # Extract all anchor tags
        links = soup.find_all('a')

        # Extract href attribute from each anchor tag
        hyperlinks = [link.get('href') for link in links if link.get('href')]

        return hyperlinks

    except requests.exceptions.RequestException as e:
        print(f"Error fetching the URL: {e}")
        return []
```

```python
if __name__ == "__main__":
    url = input("Enter the website URL (e.g., https://example.com): ")
    hyperlinks = fetch_hyperlinks(url)

    if hyperlinks:
        print("\nHyperlinks found:")
        for link in hyperlinks:
            print(link)
    else:
        print("No hyperlinks found or an error occurred.")
```

```python
from pytube import Playlist
```

```python

# Playlist URL
playlist_url = "https://www.youtube.com/playlist?list=PLxxA5z-8B2xk4szCgFmgonNcCboyNneMD"

# Create a Playlist object
playlist = Playlist(playlist_url)

print(f"Downloading videos from playlist: {playlist.title}")

# Download each video in the playlist
for video in playlist.videos:
    print(f"Downloading: {video.title}")
    try:
        video.streams.filter(progressive=True, file_extension='mp4').order_by('resolution').desc().first().download()
        print(f"Downloaded: {video.title}")
    except Exception as e:
        print(f"Failed to download {video.title}: {e}")

print("All downloads completed.")
```
